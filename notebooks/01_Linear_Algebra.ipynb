## 1. Vectors: The Building Blocks

A vector is simply a list of numbers. In ML, vectors can represent:
- Features of a data point (height, weight, age)
- Word embeddings
- Network activations

Let's visualize vectors in 2D and 3D!

```python
def plot_vector(ax, origin, vector, color='blue', label=None):
    """Plot a beautiful vector"""
    ax.quiver(origin[0], origin[1], vector[0], vector[1], 
              angles='xy', scale_units='xy', scale=1, 
              color=color, width=0.006, label=label)
    ax.plot([origin[0], origin[0] + vector[0]], 
            [origin[1], origin[1] + vector[1]], 
            'o', color=color, markersize=4)

# Create some vectors
v1 = np.array([3, 2])
v2 = np.array([1, 3])

# Visualize
fig, ax = plt.subplots(figsize=(8, 8))
plot_vector(ax, [0, 0], v1, 'blue', 'v1 = [3, 2]')
plot_vector(ax, [0, 0], v2, 'red', 'v2 = [1, 3]')

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('Vectors in 2D Space', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()

print(f"v1 = {v1}")
print(f"v2 = {v2}")
```

### Vector Operations

Let's explore vector addition, scalar multiplication, and see them visually!

```python
# Vector addition
v_sum = v1 + v2

fig, ax = plt.subplots(figsize=(10, 8))
plot_vector(ax, [0, 0], v1, 'blue', 'v1')
plot_vector(ax, [0, 0], v2, 'red', 'v2')
plot_vector(ax, [0, 0], v_sum, 'green', 'v1 + v2')

# Show the parallelogram rule
plot_vector(ax, v1, v2, 'red', alpha=0.3, linestyle='--')
plot_vector(ax, v2, v1, 'blue', alpha=0.3, linestyle='--')

ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('Vector Addition (Parallelogram Rule)', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()

print(f"v1 + v2 = {v1} + {v2} = {v_sum}")
```

```python
# Scalar multiplication
scalars = [0.5, 1, 1.5, 2]
colors = ['lightblue', 'blue', 'darkblue', 'navy']

fig, ax = plt.subplots(figsize=(10, 8))
for scalar, color in zip(scalars, colors):
    v_scaled = scalar * v1
    plot_vector(ax, [0, 0], v_scaled, color, f'{scalar} × v1')

ax.set_xlim(-1, 7)
ax.set_ylim(-1, 5)
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('Scalar Multiplication: Scaling Vectors', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()
```

### Dot Product: Measuring Similarity

The dot product is crucial in ML! It measures how similar two vectors are.

$$\mathbf{v_1} \cdot \mathbf{v_2} = |\mathbf{v_1}| |\mathbf{v_2}| \cos(\theta)$$

Where $\theta$ is the angle between the vectors.

```python
# Calculate dot product
dot_product = np.dot(v1, v2)

# Calculate angle
magnitude_v1 = np.linalg.norm(v1)
magnitude_v2 = np.linalg.norm(v2)
cos_angle = dot_product / (magnitude_v1 * magnitude_v2)
angle_rad = np.arccos(cos_angle)
angle_deg = np.degrees(angle_rad)

print(f"Dot Product: {dot_product}")
print(f"Angle between vectors: {angle_deg:.2f}°")
print(f"\nInterpretation:")
print(f"- Angle < 90°: Vectors point in similar direction")
print(f"- Angle = 90°: Vectors are orthogonal (perpendicular)")
print(f"- Angle > 90°: Vectors point in opposite directions")

# Visualize with angle
fig, ax = plt.subplots(figsize=(10, 8))
plot_vector(ax, [0, 0], v1, 'blue', 'v1')
plot_vector(ax, [0, 0], v2, 'red', 'v2')

# Draw angle arc
from matplotlib.patches import Arc
arc = Arc((0, 0), 2, 2, angle=0, theta1=0, theta2=angle_deg, color='green', linewidth=2)
ax.add_patch(arc)
ax.text(1, 0.5, f'{angle_deg:.1f}°', fontsize=12, color='green', fontweight='bold')

ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title(f'Dot Product = {dot_product}', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()
```

## 2. Matrices: Transforming Space

Matrices are transformations! They can:
- Rotate images
- Scale features
- Project data to lower dimensions

In neural networks, each layer is essentially a matrix transformation!

```python
# Create a rotation matrix
angle = np.pi / 4  # 45 degrees
rotation_matrix = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle), np.cos(angle)]
])

# Create a scaling matrix
scaling_matrix = np.array([
    [2, 0],
    [0, 0.5]
])

print("Rotation Matrix (45°):")
print(rotation_matrix)
print("\nScaling Matrix:")
print(scaling_matrix)
```

```python
# Visualize matrix transformations
def plot_transformation(matrix, title):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))
    
    # Original grid
    grid_points = []
    for x in np.linspace(-2, 2, 5):
        for y in np.linspace(-2, 2, 5):
            grid_points.append([x, y])
    grid_points = np.array(grid_points).T
    
    # Transform grid
    transformed_points = matrix @ grid_points
    
    # Plot original
    ax1.scatter(grid_points[0], grid_points[1], c='blue', s=50, alpha=0.6)
    ax1.set_xlim(-4, 4)
    ax1.set_ylim(-4, 4)
    ax1.set_xlabel('X', fontsize=12)
    ax1.set_ylabel('Y', fontsize=12)
    ax1.set_title('Original Space', fontsize=13, fontweight='bold')
    ax1.grid(True, alpha=0.3)
    ax1.axhline(y=0, color='k', linewidth=0.5)
    ax1.axvline(x=0, color='k', linewidth=0.5)
    
    # Plot unit vectors
    plot_vector(ax1, [0, 0], [1, 0], 'red', 'i')
    plot_vector(ax1, [0, 0], [0, 1], 'green', 'j')
    ax1.legend()
    
    # Plot transformed
    ax2.scatter(transformed_points[0], transformed_points[1], c='red', s=50, alpha=0.6)
    ax2.set_xlim(-4, 4)
    ax2.set_ylim(-4, 4)
    ax2.set_xlabel('X', fontsize=12)
    ax2.set_ylabel('Y', fontsize=12)
    ax2.set_title('Transformed Space', fontsize=13, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.axhline(y=0, color='k', linewidth=0.5)
    ax2.axvline(x=0, color='k', linewidth=0.5)
    
    # Plot transformed unit vectors
    new_i = matrix @ np.array([1, 0])
    new_j = matrix @ np.array([0, 1])
    plot_vector(ax2, [0, 0], new_i, 'red', "i'")
    plot_vector(ax2, [0, 0], new_j, 'green', "j'")
    ax2.legend()
    
    plt.suptitle(title, fontsize=15, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

plot_transformation(rotation_matrix, '🔄 Rotation Transformation (45°)')
plot_transformation(scaling_matrix, '📏 Scaling Transformation')
```

## 3. Eigenvalues and Eigenvectors

**Eigenvectors** are special directions that don't change direction when transformed by a matrix - they only get scaled!

$$A\mathbf{v} = \lambda\mathbf{v}$$

Where:
- $\mathbf{v}$ is an eigenvector
- $\lambda$ is the eigenvalue (scaling factor)

**Applications in ML:**
- PCA (dimensionality reduction)
- PageRank algorithm
- Understanding neural network dynamics

```python
# Create a matrix
A = np.array([[3, 1],
              [1, 3]])

# Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:")
print(A)
print(f"\nEigenvalue 1: {eigenvalues[0]:.2f}")
print(f"Eigenvector 1: {eigenvectors[:, 0]}")
print(f"\nEigenvalue 2: {eigenvalues[1]:.2f}")
print(f"Eigenvector 2: {eigenvectors[:, 1]}")
```

```python
# Visualize eigenvectors
fig, ax = plt.subplots(figsize=(10, 10))

# Plot some random vectors and their transformations
np.random.seed(42)
random_vectors = np.random.randn(2, 10) * 0.5
transformed_vectors = A @ random_vectors

for i in range(random_vectors.shape[1]):
    plot_vector(ax, [0, 0], random_vectors[:, i], 'lightblue', alpha=0.3)
    plot_vector(ax, [0, 0], transformed_vectors[:, i], 'lightcoral', alpha=0.3)

# Plot eigenvectors
eigen1 = eigenvectors[:, 0]
eigen2 = eigenvectors[:, 1]

plot_vector(ax, [0, 0], eigen1, 'blue', f'Eigenvector 1 (λ={eigenvalues[0]:.2f})')
plot_vector(ax, [0, 0], A @ eigen1, 'darkblue', 'After transformation', linestyle='--')

plot_vector(ax, [0, 0], eigen2, 'red', f'Eigenvector 2 (λ={eigenvalues[1]:.2f})')
plot_vector(ax, [0, 0], A @ eigen2, 'darkred', 'After transformation', linestyle='--')

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('Eigenvectors: Special Directions That Only Scale!', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()

print("Notice: Eigenvectors (solid) and their transformations (dashed) point in the same direction!")
print("They only get scaled by their eigenvalues.")
```

## 4. ML Application: Principal Component Analysis (PCA)

PCA uses eigenvectors to find the most important directions in your data!

```python
# Generate correlated data
np.random.seed(42)
mean = [0, 0]
cov = [[3, 1.5], [1.5, 1]]  # Covariance matrix
data = np.random.multivariate_normal(mean, cov, 500).T

# Compute PCA (eigenvectors of covariance matrix)
eigenvalues_pca, eigenvectors_pca = np.linalg.eig(cov)

# Sort by eigenvalue
idx = eigenvalues_pca.argsort()[::-1]
eigenvalues_pca = eigenvalues_pca[idx]
eigenvectors_pca = eigenvectors_pca[:, idx]

# Visualize
fig, ax = plt.subplots(figsize=(12, 10))

# Plot data
ax.scatter(data[0], data[1], alpha=0.5, s=30, label='Data points')

# Plot principal components
scale = 3
pc1 = eigenvectors_pca[:, 0] * scale * np.sqrt(eigenvalues_pca[0])
pc2 = eigenvectors_pca[:, 1] * scale * np.sqrt(eigenvalues_pca[1])

plot_vector(ax, [0, 0], pc1, 'red', f'PC1 (explains {eigenvalues_pca[0]/sum(eigenvalues_pca)*100:.1f}%)')
plot_vector(ax, [0, 0], -pc1, 'red')
plot_vector(ax, [0, 0], pc2, 'green', f'PC2 (explains {eigenvalues_pca[1]/sum(eigenvalues_pca)*100:.1f}%)')
plot_vector(ax, [0, 0], -pc2, 'green')

ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_title('PCA: Finding Important Directions with Eigenvectors', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
ax.axis('equal')
plt.show()

print("PC1 (red) points in the direction of maximum variance!")
print("PC2 (green) is orthogonal to PC1 and captures remaining variance.")
```
