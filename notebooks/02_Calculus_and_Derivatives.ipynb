## 1. What is a Derivative?

A derivative tells us **how fast something is changing**.

$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

Geometrically, it's the **slope of the tangent line** at a point!

```python
# Define a function
def f(x):
    return 0.5 * x**2 + x + 1

# Analytical derivative
def f_prime(x):
    return x + 1

# Visualize function and tangent line
def plot_tangent(x0):
    x = np.linspace(-5, 5, 200)
    y = f(x)
    
    # Tangent line at x0
    slope = f_prime(x0)
    y_tangent = slope * (x - x0) + f(x0)
    
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Plot function
    ax.plot(x, y, 'b-', linewidth=2.5, label=f'f(x) = 0.5x² + x + 1')
    
    # Plot tangent line
    ax.plot(x, y_tangent, 'r--', linewidth=2, label=f'Tangent at x={x0}')
    
    # Mark the point
    ax.plot(x0, f(x0), 'ro', markersize=12, label=f'Point ({x0}, {f(x0):.2f})')
    
    # Add slope annotation
    ax.annotate(f'Slope = {slope:.2f}', 
                xy=(x0, f(x0)), 
                xytext=(x0+1.5, f(x0)+2),
                fontsize=13,
                fontweight='bold',
                arrowprops=dict(arrowstyle='->', lw=2, color='red'))
    
    ax.set_xlabel('x', fontsize=13)
    ax.set_ylabel('f(x)', fontsize=13)
    ax.set_title('Derivative = Slope of Tangent Line', fontsize=15, fontweight='bold')
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3)
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    plt.show()

# Try different points
plot_tangent(2)
print(f"At x=2, the derivative (slope) is: {f_prime(2)}")
```

### Interactive: See How Tangent Changes!

```python
# Show multiple tangent lines
fig, ax = plt.subplots(figsize=(14, 9))

x = np.linspace(-5, 5, 200)
y = f(x)
ax.plot(x, y, 'b-', linewidth=3, label='f(x)', alpha=0.8)

# Multiple points
x_points = [-3, -1, 0, 2, 4]
colors = ['red', 'green', 'orange', 'purple', 'brown']

for x0, color in zip(x_points, colors):
    slope = f_prime(x0)
    y_tangent = slope * (x - x0) + f(x0)
    ax.plot(x, y_tangent, '--', color=color, linewidth=1.5, alpha=0.7)
    ax.plot(x0, f(x0), 'o', color=color, markersize=10)
    ax.text(x0, f(x0) - 1.5, f"m={slope:.1f}", fontsize=11, ha='center', 
            fontweight='bold', color=color)

ax.set_xlabel('x', fontsize=13)
ax.set_ylabel('f(x)', fontsize=13)
ax.set_title('How Derivative Changes Across the Function', fontsize=15, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
plt.show()

print("Notice: The slope changes at different points!")
print("Negative slope → function decreasing")
print("Positive slope → function increasing")
print("Zero slope → local minimum or maximum")
```

## 2. Computing Derivatives

### Method 1: Numerical Approximation

```python
def numerical_derivative(f, x, h=1e-5):
    """Compute derivative numerically using finite differences"""
    return (f(x + h) - f(x)) / h

# Test it
x_test = 2.0
numerical = numerical_derivative(f, x_test)
analytical = f_prime(x_test)

print(f"At x = {x_test}:")
print(f"Numerical derivative: {numerical:.6f}")
print(f"Analytical derivative: {analytical:.6f}")
print(f"Difference: {abs(numerical - analytical):.2e}")
print("\nThey're almost identical!")
```

### Visualizing Numerical Approximation

```python
# Show how h affects approximation
x0 = 2
h_values = [2, 1, 0.5, 0.1]

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.ravel()

x = np.linspace(-1, 5, 200)
y = f(x)

for idx, h in enumerate(h_values):
    ax = axes[idx]
    
    # Plot function
    ax.plot(x, y, 'b-', linewidth=2.5, label='f(x)', alpha=0.7)
    
    # Secant line
    slope_approx = (f(x0 + h) - f(x0)) / h
    y_secant = slope_approx * (x - x0) + f(x0)
    ax.plot(x, y_secant, 'r--', linewidth=2, label=f'Secant (h={h})')
    
    # True tangent
    slope_true = f_prime(x0)
    y_tangent = slope_true * (x - x0) + f(x0)
    ax.plot(x, y_tangent, 'g:', linewidth=2, label='True tangent', alpha=0.7)
    
    # Mark points
    ax.plot([x0, x0+h], [f(x0), f(x0+h)], 'ro', markersize=8)
    
    # Show h
    ax.plot([x0, x0+h], [f(x0), f(x0)], 'k-', linewidth=1.5)
    ax.text(x0+h/2, f(x0)-0.5, f'h={h}', fontsize=11, ha='center')
    
    ax.set_xlim(-1, 5)
    ax.set_ylim(-2, 12)
    ax.set_xlabel('x', fontsize=11)
    ax.set_ylabel('f(x)', fontsize=11)
    ax.set_title(f'h = {h}, Slope ≈ {slope_approx:.3f} (True: {slope_true:.3f})', 
                 fontsize=12, fontweight='bold')
    ax.legend(fontsize=10)
    ax.grid(True, alpha=0.3)

plt.suptitle('As h → 0, Secant Line → Tangent Line', fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()
```

## 3. Common Derivative Rules

### Power Rule
$$\frac{d}{dx}x^n = nx^{n-1}$$

### Product Rule
$$\frac{d}{dx}[f(x)g(x)] = f'(x)g(x) + f(x)g'(x)$$

### Chain Rule (MOST IMPORTANT for ML!)
$$\frac{d}{dx}f(g(x)) = f'(g(x)) \cdot g'(x)$$

```python
# Visualize chain rule
def outer_function(u):
    return u**2

def inner_function(x):
    return 2*x + 1

def composed_function(x):
    return outer_function(inner_function(x))

# Derivatives
def d_outer(u):
    return 2*u

def d_inner(x):
    return 2

def d_composed(x):
    # Chain rule: f'(g(x)) * g'(x)
    return d_outer(inner_function(x)) * d_inner(x)

x = np.linspace(-3, 3, 200)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Inner function
axes[0].plot(x, inner_function(x), 'b-', linewidth=2.5)
axes[0].set_title('Inner: g(x) = 2x + 1', fontsize=13, fontweight='bold')
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('g(x)', fontsize=12)
axes[0].grid(True, alpha=0.3)
axes[0].text(0, -3, "g'(x) = 2", fontsize=12, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Outer function
u = np.linspace(-5, 5, 200)
axes[1].plot(u, outer_function(u), 'g-', linewidth=2.5)
axes[1].set_title('Outer: f(u) = u²', fontsize=13, fontweight='bold')
axes[1].set_xlabel('u', fontsize=12)
axes[1].set_ylabel('f(u)', fontsize=12)
axes[1].grid(True, alpha=0.3)
axes[1].text(0, 15, "f'(u) = 2u", fontsize=12, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Composed function
axes[2].plot(x, composed_function(x), 'r-', linewidth=2.5)
axes[2].set_title('Composed: f(g(x)) = (2x+1)²', fontsize=13, fontweight='bold')
axes[2].set_xlabel('x', fontsize=12)
axes[2].set_ylabel('f(g(x))', fontsize=12)
axes[2].grid(True, alpha=0.3)
axes[2].text(0, 25, "(f∘g)'(x) = f'(g(x))·g'(x)\n= 2(2x+1)·2\n= 4(2x+1)", 
             fontsize=11, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

plt.suptitle('🔗 Chain Rule: Composition of Functions', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# Verify numerically
x_test = 1.5
print(f"\nAt x = {x_test}:")
print(f"Chain rule derivative: {d_composed(x_test)}")
print(f"Numerical derivative: {numerical_derivative(composed_function, x_test):.6f}")
```

## 4. Derivatives of Common Activation Functions

Neural networks use non-linear activation functions. Let's visualize them and their derivatives!

```python
# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# Visualize
x = np.linspace(-5, 5, 500)

fig, axes = plt.subplots(3, 2, figsize=(16, 14))

activations = [
    (sigmoid, sigmoid_derivative, 'Sigmoid σ(x) = 1/(1+e⁻ˣ)'),
    (tanh, tanh_derivative, 'Tanh tanh(x)'),
    (relu, relu_derivative, 'ReLU max(0,x)')
]

for idx, (func, deriv, name) in enumerate(activations):
    # Function
    axes[idx, 0].plot(x, func(x), 'b-', linewidth=3)
    axes[idx, 0].set_title(f'{name}', fontsize=13, fontweight='bold')
    axes[idx, 0].set_xlabel('x', fontsize=12)
    axes[idx, 0].set_ylabel('f(x)', fontsize=12)
    axes[idx, 0].grid(True, alpha=0.3)
    axes[idx, 0].axhline(y=0, color='k', linewidth=0.5)
    axes[idx, 0].axvline(x=0, color='k', linewidth=0.5)
    
    # Derivative
    axes[idx, 1].plot(x, deriv(x), 'r-', linewidth=3)
    axes[idx, 1].set_title(f"Derivative of {name.split()[0]}", fontsize=13, fontweight='bold')
    axes[idx, 1].set_xlabel('x', fontsize=12)
    axes[idx, 1].set_ylabel("f'(x)", fontsize=12)
    axes[idx, 1].grid(True, alpha=0.3)
    axes[idx, 1].axhline(y=0, color='k', linewidth=0.5)
    axes[idx, 1].axvline(x=0, color='k', linewidth=0.5)

plt.suptitle('🧠 Neural Network Activation Functions & Their Derivatives', 
             fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()

print("Key observations:")
print("• Sigmoid: Derivative is small for large |x| → vanishing gradient problem")
print("• Tanh: Similar to sigmoid but centered at 0")
print("• ReLU: Derivative is 0 for x<0, 1 for x>0 → simple and fast!")
```

## 5. Finding Minima: Where Derivatives = 0

In ML, we want to minimize loss functions. At a minimum, the derivative is zero!

```python
# A function with a minimum
def loss_function(x):
    return (x - 2)**2 + 1

def loss_derivative(x):
    return 2*(x - 2)

x = np.linspace(-2, 6, 300)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 12))

# Loss function
ax1.plot(x, loss_function(x), 'b-', linewidth=3, label='Loss L(x)')
ax1.plot(2, loss_function(2), 'r*', markersize=25, label='Minimum at x=2')
ax1.set_xlabel('x (parameter)', fontsize=13)
ax1.set_ylabel('Loss', fontsize=13)
ax1.set_title('Loss Function', fontsize=14, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.axhline(y=loss_function(2), color='r', linestyle='--', alpha=0.3)

# Derivative
ax2.plot(x, loss_derivative(x), 'g-', linewidth=3, label="Derivative L'(x)")
ax2.plot(2, 0, 'r*', markersize=25, label='Zero at x=2')
ax2.axhline(y=0, color='k', linewidth=1.5)
ax2.axvline(x=2, color='r', linestyle='--', alpha=0.5)
ax2.set_xlabel('x (parameter)', fontsize=13)
ax2.set_ylabel('Derivative', fontsize=13)
ax2.set_title('Derivative of Loss', fontsize=14, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)

# Add annotations
ax2.annotate('Negative slope\n← Go right', xy=(-1, loss_derivative(-1)), 
             xytext=(-1, -8), fontsize=12, ha='center',
             arrowprops=dict(arrowstyle='->', lw=2, color='blue'))
ax2.annotate('Positive slope\n→ Go left', xy=(5, loss_derivative(5)), 
             xytext=(5, 8), fontsize=12, ha='center',
             arrowprops=dict(arrowstyle='->', lw=2, color='blue'))

plt.suptitle('🎯 Finding Minimum: Where Derivative = 0', fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()

print("This is the essence of optimization in ML!")
print("We follow the negative gradient to find the minimum loss.")
```

## 6. Partial Derivatives

When we have multiple variables (which we always do in ML!), we use **partial derivatives**.

```python
# Function of two variables
def f_2d(x, y):
    return x**2 + y**2

# Partial derivatives
def df_dx(x, y):
    return 2*x

def df_dy(x, y):
    return 2*y

# Create meshgrid
x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = f_2d(X, Y)

# 3D plot
fig = plt.figure(figsize=(16, 6))

# Surface plot
ax1 = fig.add_subplot(121, projection='3d')
surf = ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8, edgecolor='none')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('y', fontsize=12)
ax1.set_zlabel('f(x,y)', fontsize=12)
ax1.set_title('Function: f(x,y) = x² + y²', fontsize=13, fontweight='bold')
fig.colorbar(surf, ax=ax1, shrink=0.5)

# Contour plot with gradient vectors
ax2 = fig.add_subplot(122)
contour = ax2.contour(X, Y, Z, levels=15, cmap='viridis')
ax2.clabel(contour, inline=True, fontsize=10)

# Add gradient vectors
step = 5
for i in range(0, X.shape[0], step):
    for j in range(0, X.shape[1], step):
        dx = df_dx(X[i,j], Y[i,j])
        dy = df_dy(X[i,j], Y[i,j])
        ax2.arrow(X[i,j], Y[i,j], -dx*0.1, -dy*0.1, 
                 head_width=0.15, head_length=0.1, fc='red', ec='red', alpha=0.6)

ax2.set_xlabel('x', fontsize=12)
ax2.set_ylabel('y', fontsize=12)
ax2.set_title('Contours & Gradient Vectors', fontsize=13, fontweight='bold')
ax2.set_aspect('equal')
ax2.plot(0, 0, 'r*', markersize=20, label='Minimum')
ax2.legend(fontsize=11)

plt.suptitle('📊 Partial Derivatives in 2D', fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()

print("Red arrows show negative gradient direction → toward minimum")
print(f"∂f/∂x = 2x")
print(f"∂f/∂y = 2y")
print(f"At (1,1): gradient = [{df_dx(1,1)}, {df_dy(1,1)}]")
```

## 🚀 Key Takeaways

1. **Derivatives measure rate of change** - the slope of the tangent line
2. **Chain rule** is the foundation of backpropagation
3. **Activation function derivatives** are computed during training
4. **Zero derivatives** indicate potential minima (or maxima/saddle points)
5. **Partial derivatives** extend to multiple variables
6. **Gradient** = vector of all partial derivatives → points toward steepest ascent
